```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

# 1. Load the data

```python
df = pd.read_csv('\data\raw\train.csv', 
                 parse_dates=['Date'], 
                 low_memory=False)

store = pd.read_csv('\data\raw\store.csv')

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
```

# Question 1: Any missing values?

```python
print("\n1. Missing values per column:")
print(df.isnull().sum())

# Quick summary
if df.isnull().sum().sum() == 0:
    print("→ No missing values in the train dataset.")
else:
    print("→ There ARE missing values — see counts above.")
```

# Question 2: Date range

```python
print("\n2. Date range:")
print("Earliest date:", df['Date'].min().date())
print("Latest date:  ", df['Date'].max().date())
print("Total unique days:", df['Date'].nunique())
```

# Question 3: How many stores?

```python
print("\n3. Number of unique stores:", df['Store'].nunique())

# Optional: stores per day (should be roughly constant)
print("Average records per day:", len(df) / df['Date'].nunique())
```

# Question 4: Distribution of sales

```python
print("\n4. Sales descriptive statistics (all days):")
print(df['Sales'].describe())

# Most useful view: sales only when store is open
sales_open = df[df['Open'] == 1]['Sales']

print("\nSales statistics (only when Open == 1):")
print(sales_open.describe())

# Visualization
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(sales_open, bins=50, kde=True, color='royalblue')
plt.title('Sales Distribution (Open = 1)')
plt.xlabel('Sales')

plt.subplot(1, 2, 2)
sns.boxplot(x=sales_open, color='lightgreen')
plt.title('Sales Boxplot (Open = 1)')

plt.tight_layout()
plt.show()

# Optional: log scale view (helps see skewness)
plt.figure(figsize=(10, 4))
sns.histplot(np.log1p(sales_open), bins=50, kde=True, color='purple')
plt.title('Log(Sales + 1) Distribution (Open = 1)')
plt.xlabel('log(Sales + 1)')
plt.show()
```

# Question 5: Sales when Open == 0

```python

print("\n5. Sales when Open == 0:")
sales_closed = df[df['Open'] == 0]['Sales']

print("Unique sales values when closed:", sales_closed.unique())
print("Count of records where Open == 0:", len(sales_closed))
print("Percentage of closed days:", 
      100 * len(sales_closed) / len(df), "%")

# Should always be zero — sanity check
if (sales_closed == 0).all():
    print("→ All sales are 0 when store is closed (as expected).")
else:
    print("→ WARNING: some sales > 0 when store marked closed!")
```
